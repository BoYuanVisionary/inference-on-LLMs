```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '2,3'
from reasoning.models.model import ValueModel_qwen

value_model = ValueModel_qwen(device = "auto")
from reasoning.models.model_vllm import QwenPolicy
policy_model = QwenPolicy('Qwen/Qwen2.5-7B-Instruct', gpu_memory_utilization=0.8, max_new_tokens=2048)
```

```python
from reasoning.SamplingTree.path import Path
question = "Find the greatest integer less than $(\\sqrt{7} + \\sqrt{5})^6.$  (Do not use a calculator!)"
solutions = """
To find the greatest integer less than \\((\\sqrt{7} + \\sqrt{5})^6\\), we will use a combination of algebraic manipulation and estimation. Let's denote \\( x = \\sqrt{7} + \\sqrt{5} \\). We need to find the greatest integer less than \\( x^6 \\).\n\nFirst, consider the conjugate of \\( x \\), which is \\( y = \\sqrt{7} - \\sqrt{5} \\). Note that \\( 0 < y < 1 \\) because \\( \\sqrt{7} > \\sqrt{5} \\).\n\nNext, we calculate \\( x + y \\):\n\\[\nx + y = (\\sqrt{7} + \\sqrt{5}) + (\\sqrt{7} - \\sqrt{5}) = 2\\sqrt{7}\n\\]\n\nNow, we calculate \\( x - y \\):\n\\[\nx - y = (\\sqrt{7} + \\sqrt{5}) - (\\sqrt{7} - \\sqrt{5}) = 2\\sqrt{5}\n\\]\n\nWe will use the binomial theorem to expand \\( (x + y)^6 \\) and \\( (x - y)^6 \\):\n\\[\n(x + y)^6 = (2\\sqrt{7})^6 = 2^6 \\cdot (\\sqrt{7})^6 = 64 \\cdot 7^3 = 64 \\cdot 343 = 21952\n\\]\n\n\\[\n(x - y)^6 = (2\\sqrt{5})^6 = 2^6 \\cdot (\\sqrt{5})^6 = 64 \\cdot 5^3 = 64 \\cdot 125 = 8000\n\\]\n\nSince \\( 0 < y < 1 \\), \\( (x - y)^6 \\) is a very small positive number. Therefore, \\( x^6 \\) is slightly less than \\( 21952 \\).\n\nTo find the exact value of \\( x^6 \\), we use the fact that:\n\\[\nx^6 + (x - y)^6 = (x + y)^6\n\\]\n\nGiven \\( (x - y)^6 \\) is very small, we approximate:\n\\[\nx^6 \\approx 21952 - (x - y)^6\n\\]\n\nSince \\( (x - y)^6 \\) is a very small number, we can estimate that \\( x^6 \\) is just slightly less than 21952. To find the greatest integer less than \\( x^6 \\), we conclude:\n\\[\n\\boxed{21951}\n\\]"""
path = Path(solutions)
for step in path.steps:
    print("-"*100)
    print(step)
```

```python
scores = value_model.get_value_with_steps(question, path.steps)

print(scores)
```

```python

system_prompt = "Please reason step by step and put the final answer in \\boxed{}"

for i in [1,2,3,4,5,6,7]:
    partial_solutions =' '.join(path.steps[:i])
    user_prompt = question 
    partial_solutions + '\n Wait, wait. Wait. Let’s reevaluate this step-by-step. '

    outputs = policy_model.get_proposal(system_prompt, user_prompt, n_responses=1, partial_solutions=partial_solutions)

    print(user_prompt)

    print('begin of output')
    for output in outputs:

        print(output)
        print("-"*100)


```
